```python
import numpy as np
import torch
```

```python
arr = np.array([1.0, 2, 3, 5])

x = torch.from_numpy

print('numpy 배열:', arr)
print('tensor :', x)

arr[0] = 100
print('변경후')
print('numpy 배열 :', arr)
print('tensor :', x)
```

-- torch.tensor(array) : 데이터를  복사, 원본과 분리된 tensor생성

```python
arr = np.array([1.0,2,3,4])
x=torch.tensor(arr)

print('변경전')
print('numpy 배열 :', arr)
print('tensor :', x)

arr[0] = 100
print('변경후')
print('numpy 배열 :', arr)
print('tensor :', x)
```

- torch.as_tensor(array)
    :가능하면 메모리를 공유하고 필요한 경우에만 데이터를 복사하는 함수
    -torch.form_numpy()와 비슷하게 동작하지만, 입력 자료형이나, 장치(device)등을 지정할수있다는 차이

```python
arr = np.array([1.0,2,3,4])
x=torch.as_tensor(arr)

print('변경전')
print('numpy 배열 :', arr)
print('tensor :', x)

arr[0] = 100
print('변경후')
print('numpy 배열 :', arr)
print('tensor :', x)
```

```python
arr = np.array([1,2,3,4], dtype=np.int32)
x=torch.as_tensor(arr, dtype=torch.float32)

print('변경전')
print('numpy 배열 :', arr)
print('tensor :', x)

arr[0] = 100
print('변경후')
print('numpy 배열 :', arr)
print('tensor :', x)
```

#### - PyTorch Tensor를 NumPy배열로 변환
: 모델이 계산한 결과를 python의 다른 라이브러리로 전달하거나, 데이터를 시각화하고 분석할 때 사용

```python
x = torch.tensor([0.1,0.2,0.3,0.4])

arr = x.numpy()

print('tensor :', x)
print('tensor :', x)
print('자료형 :', arr.dtype)
print('크기 :', arr.shape)

x[0] = 0.7
print('변경후')
print('arr :', arr)
print('x :', x)
```
